## E1. String manipulation
Given `s = " Big Data Tools, Python and Big Data Analytics "`:
- Remove the surrounding whitespace and convert to lower and upper case.
- Check whether it starts with "big" and ends with "analytics".
- Find the index of the first and the second "Big"
  - *hint: find accepts a start position.*
- Replace "Big Data" with "BD".
- Split the string by "," and rejoin the pieces using " | ".

```python
s = " Big Data Tools, Python and Big Data Analytics "

# 1) strip() removes leading/trailing whitespace; strings are immutable,
#    so every method returns a NEW string that we store in a variable
clean = s.strip()
lower = clean.lower()
upper = clean.upper()
print("Stripped:", repr(clean))
print("Lower   :", lower)
print("Upper   :", upper)

# 2) startswith/endswith are case sensitive, so we test on the lowercase version
print("Starts with 'big'      :", lower.startswith("big"))
print("Ends with 'analytics'  :", lower.endswith("analytics"))

# 3) find() returns the index of the first occurrence (or -1 if not found).
#    For the second occurrence we restart the search just after the first one
#    by passing a start position as second argument.
first = clean.find("Big")
second = clean.find("Big", first + 1)
print("First 'Big' index :", first)
print("Second 'Big' index:", second)

# 4) replace() substitutes ALL occurrences of the old substring
replaced = clean.replace("Big Data", "BD")
print("Replaced:", replaced)

# 5) split(",") returns a list of items;
#    join(" | ") returns a string where each item has been condensed with " | " and the others
items = clean.split(",")
print("Items:", items)
print("Joined:", " | ".join(items))
```

## E2. List manipulation
- Start from `a = ['Raj', 'Penny', 'Sheldon']`
- Obtain `['Bernardette', 'Penny', 'Howard', 'Amy', 'Sheldon']`, using pop, append, insert, extend, remove, and index at least once each.
- Print the list reversed using slicing, and every element with an even index.

```python
a = ['Raj', 'Penny', 'Sheldon']

print(f"Original list: {a}")

# pop(idx) removes (and returns) the element at position idx
# pop(0) removes first element (or with index 0) -> 'Raj' leaves the list
# pop() by default removes the last element
removed = a.pop(0)
print(f"pop: {removed} -> {a}")

# insert(0, x) puts 'Bernardette' in the first position
a.insert(0, 'Bernardette')
print(f"insert : {a}")

# extend() adds all the elements of another list at the end
a.extend(['Howard', 'Amy'])
print(f"extend : {a}")      # Sheldon is now in the wrong place (not last)

# Sheldon must be last: two options
# OPTION 1
# remove() deletes the first matching element,
# then append() adds it again at the end of the list
# a.remove('Sheldon')
# a.append('Sheldon')
# print(f"remove + append: {a}")

# OR
# OPTION 2
# index(elemnt) returns the position of elemnt:
# pop(idx) deletes element at position idx
# then append() adds it again at the end of the list
idx = a.index('Sheldon')
idx = a.pop(idx) 
a.append('Sheldon')
print(f"index + pop + append: {a}")

# Verify that we obtained the requested list
print(f"Is the obtained list equal to the desired one? {a == ['Bernardette', 'Penny', 'Howard', 'Amy', 'Sheldon']}")

# Slicing [start:stop:step]
print("Reversed        :", a[::-1])   # step -1 walks the list backwards
print("Even indexes    :", a[::2])    # indexes 0, 2, 4, ...
```

## E3. Statistical measures
- Calculate mean, variance and standard deviations of a given list of numbers
- Double check the obtained results with native functions in statistics module

```python
import math
import statistics

data = [24, 25, 27, 18, 20, 22, 23, 27, 30]   # sample list of numbers
n = len(data)

# Mean = sum of the values / number of values
mean = sum(data) / n

# Sum of squared deviations from the mean
sq_dev = 0

# with for
for x in data:
      sq_dev += (x - mean)**2

# condensed version
#sq_dev = sum((x - mean) ** 2 for x in data)

# Population variance divides by n; sample variance divides by n - 1
pop_var = sq_dev / n
sam_var = sq_dev / (n - 1)

# Standard deviation = square root of the variance
pop_std = math.sqrt(pop_var)
sam_std = math.sqrt(sam_var)

print(f"Manual     -> mean={mean:.4f}  pvar={pop_var:.4f}  pstd={pop_std:.4f}  var={sam_var:.4f}  std={sam_std:.4f}")

# Check with the native statistics module
print(f"statistics -> mean={statistics.mean(data):.4f}  pvar={statistics.pvariance(data):.4f}  pstd={statistics.pstdev(data):.4f}  var={statistics.variance(data):.4f}  std={statistics.stdev(data):.4f}")
```

## E4. Degrees to Radians
- Implement a radians function that returns the radian equivalent of a degree. Use the following formula: `rad = degrees * pi/180`
- Use this function to print the radian equivalent of all degrees ranging from 1° to 180°. Use two digits of precision for the results.

```python
from math import pi

def radians(degrees):
    # Return the radian equivalent of an angle expressed in degrees.
    return degrees * pi / 180

# range(1, 181) -> 1, 2, ..., 180 (the stop value is excluded)
for deg in range(1, 181):
    # {value:.2f} formats a float with two digits after the decimal point
    print(f"{deg:3d}° = {radians(deg):.2f} rad")
```

## E5. Duplicate Elimination
In organizations, a list of email addresses is often compiled for marketing purposes. However, duplicate email addresses need to be removed from this list. Write a function that receives a list and returns a list containing only unique values. Test your function with a list of email addresses.

```python
def unique(values):
    res = []
    for val in values:
        if val.lower() not in res:
            res.append(val.lower())
    return res

emails = ["anna@unipa.it", "bob@example.com", "mario@unipa.it", "anna@unipa.it", "carla@example.com", "BOB@EXAMPLE.COM"]

print("Original :", emails)
print("Unique   :", unique(emails))
```

## E6. Counting Votes
Given the votes, arranged as a string where each vote is separated from the next by a comma, write a script that uses a dictionary to determine the number of votes received by candidates in an election.

```python
votes = "Alice,Bob,Alice,Carla,Bob,Alice,Carla,ALICE,Bob"

# split(",") turns the string into a list of votes
splitted_votes = votes.split(',')
print(splitted_votes)

# Dictionary: key = candidate, value = number of votes
counts = {}

#SOLUTION 1
# for vote in splitted_votes:
#     vote = vote.strip().lower()    # remove accidental spaces and normalizes to lower case
#     if not vote in counts:
#         counts[vote] = 1
#     else:
#         counts[vote] += 1

# SOLUTION 2 with get
for vote in splitted_votes:
    vote = vote.strip().lower()    # remove accidental spaces and normalizes to lower case
    # get(key, default) returns the value associated to the given key or the specified default value if the dictionary has no the given key
    counts[vote] = counts.get(vote, 0) + 1

# Print candidates sorted by number of votes (descending)
for candidate, n in sorted(counts.items(), key=lambda x: x[1], reverse=True):
    print(f"{candidate:<8} {n}")    # candidate:<8 format the text so as all candidates spread to a maximum of 8 characters, thus values n are aligned

# Winner = key with the maximum value
winner = max(counts, key=counts.get)
print("Winner:", winner)
```

## E8. Class Average – CSV
- An instructor teaches a class in which each student takes three exams. The instructor would like to store this information in a file named grades.csv. Write code that enables an instructor to enter each student's first name and last name as strings and the student's three exam grades as integers.
- Use the csv module to write each record into the grades.csv file. Each record should be a single line of text in the following CSV format: `firstname,lastname,exam1grade,exam2grade,exam3grade`
- Read the grades.csv and display the data in tabular format.

```python
import csv

def enter_student():
    # Interactive entry of ONE student 
    first = input("First name: ")
    last = input("Last name: ")
    g1 = int(input('1st grade: '))
    g2 = int(input('2nd grade: '))
    g3 = int(input('3rd grade: '))
    return [first, last, g1,g2,g3]

# Sample records (first name, last name, exam1, exam2, exam3)
# students = [
#     ["Sheldon", "Cooper", 100, 98, 95],
#     ["Leonard", "Hofstadter", 90, 92, 85],
#     ["Raj", "Koothrappali", 85, 90, 88],
# ]

with open("grades.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    
    while True:
        student = enter_student()
        writer.writerow(student) # one line of text per student

        e = str(input('continue? [Y/N] '))
        if e.lower() == 'n' or e.lower() == 'no':
            break

print('file saved')

with open("grades.csv", "r", newline="", encoding="utf-8") as f:
    reader = csv.reader(f)

    # Tabular display: <12 = left aligned in 12 chars, >6 = right aligned in 6 chars
    print(f"{'First name':<12}{'Last name':<14}{'Exam1':>6}{'Exam2':>6}{'Exam3':>6}")
    print("-" * 44)
    for record in reader:
        firstname, lastname, exam1grade, exam2grade, exam3grade = record
        print(f'{firstname:<12}{lastname:<14}{int(exam1grade):>6}{int(exam2grade):>6}{int(exam3grade):>6}')
```

- Use the json module to write the student information to the file in JSON format, create a dictionary of student data in the following format: `gradebook_dict = {'students': [student1dictionary, student2dictionary, ...]}`
- Each dictionary in the list represents one student and contains the keys 'first_name', 'last_name', 'exam1', 'exam2' and 'exam3', which map to the values representing each student's first name (string), last name (string) and three exam scores (integers).
- Output the gradebook_dict in JSON format to the file grades.json.
- Read the grades.json and display the data in tabular format.

```python
import json

# Build one dictionary per student

student_list = []

while True:
    
    student = enter_student()
    firstname, lastname, exam1grade, exam2grade, exam3grade = student
    student_list.append(student)

    e = str(input('continue? [Y/N] '))
    if e.lower() == 'n' or e.lower() == 'no':
        break

gradebook_dict = {
    "students": [
        {"first": f, "last": l, "exam1": e1, "exam2": e2, "exam3": e3}
        for f, l, e1, e2, e3 in student_list
    ]
}

# json.dump writes the dictionary to file; indent=4 makes it human readable
with open("grades.json", "w", encoding="utf-8") as f:
    json.dump(gradebook_dict, f, indent=4)

# json.load reads the file back into a Python dictionary
with open("grades.json", "r", encoding="utf-8") as f:
    loaded = json.load(f)

# Tabular display of the data read from the JSON file
print(f"{'First name':<12}{'Last name':<14}{'Exam1':>6}{'Exam2':>6}{'Exam3':>6}")
print("-" * 44)
for st in loaded["students"]:
    print(f"{st['first']:<12}{st['last']:<14}"
          f"{st['exam1']:>6}{st['exam2']:>6}{st['exam3']:>6}")
```
