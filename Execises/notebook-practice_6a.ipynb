**Insert**


```python
article_1 = {"_id": "n1",
           "author": "Derrick Mwiti",
           "about": "Introduction to MongoDB and Python",
           "tags": ["mongodb", "python", "pymongo"]}
```

```python
# Insert an item into the collection, created on the fly

result = db.articles.insert_one(article_1)
print(f'First article key is: {result.inserted_id}')
```

```python
print(db.list_collection_names())
```

```python
# Insert multiple items into the collection

article1 = {"author": "Emmanuel Kens",
            "about": "Knn and Python",
            "tags": ["Knn","pymongo"]}

article2 = {"_id": "n3", "author": "Daniel Kimeli",
            "about": "Web Development and Python",
            "tags": ["web", "design", "HTML"]}

new_articles = db.articles.insert_many([article1, article2])
print(f'The new article IDs are {new_articles.inserted_ids}')
```

```python
# prints all the articles, for each selects and show id, author and about

for doc in db.articles.find({},{ "_id": 1, "author": 1, "about": 1}):
  print(doc)
```

```python
# explicitely create a new collection 
db.create_collection('articles_new')

print(db.list_collection_names())
```

**update**

```python
# Update all documents in the 'articles' collection to set 'join_date' to 'today'

db.articles.update_many({}, { "$set":{ 'join_date': 'today'}})
```

```python
# Update all documents where Derrick Mwiti is an author

query = { "author": "Derrick Mwiti" }
new_author = { "$set": { "author": "John David" } }
db.articles.update_many(query, new_author)
```

```python
for doc in db.articles.find({},{ "_id": 0, "author": 1, "about": 1}):
  print(doc)
```

**delete**

```python
# Update all documents where John David is an author

query = { "author": "John David" }
db.articles.delete_one(query)
for doc in db.articles.find({},{ "_id": 0, "author": 1, "about": 1}):
  print(doc)
```

```python
# general delete of items in a collection

delete_articles = db.articles.delete_many({})
print(delete_articles.deleted_count, " articles deleted.")
```

**drop**

```python
# drop deletes the collection 

db.articles.drop()
print(db.list_collection_names())
```

```python
db.articles_new.drop()  # Remove the articles collection
client.drop_database('BigDataTools')  # Remove the BigDataTools database
```
