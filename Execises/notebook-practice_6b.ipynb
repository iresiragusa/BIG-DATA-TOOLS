```python
# find Italian movies in Italian and sort them by year, retrieve the first 5
query = {"countries" : "Italy", "languages" : "Italian"}
prop = {"_id":0, "title":1, "year":1, "runtime":1, "languages":1, "fullplot":1}
res = db.movies.find(query, prop)

list(res.sort({'year':-1}).limit(5))
```

```python
# find movies rated as R with Meryl Streep or Charlize Theron
query = {"$or" : [{"cast":"Meryl Streep"}, {"cast" : "Charlize Theron"}], "rated":"R"}
prop = {"_id":0, "title":1, "year":1, "cast":1, "rated":1}
res = db.movies.find(query, prop)
list(res.limit(3))
```

```python
# find movies with Charlize Theron with year <= 2015
query = {"cast" : "Charlize Theron", "year":{"$lte":2015}}
prop = {"_id":0, "title":1, "year":1, "cast":1, "rated":1}
res = db.movies.find(query, prop)
list(res.limit(3))
```

```python
# find moview that are not movie-type movie and with num_mflix_comments > 0
query = {"type": {"$ne" : "movie"}, "num_mflix_comments": {"$gt" : 0}}
prop = {"_id":0, "title":1, "type":1, "cast":1, "num_mflix_comments":1}
res = db.movies.find(query, prop)
list(res.limit(3))
```

```python
# find movies with imdb rating > 9
query = {"imdb.rating": {"$gt" : 9}}
prop = {"_id":0, "title":1, "type":1, "cast":1, "imdb":1}
res = db.movies.find(query, prop)
list(res.limit(3))
```

**Aggregations**

```python
# how many movies are in the collection?

group_stg = {"$group":{"_id":"null", "count":{"$sum":1}}}
res = db.movies.aggregate([group_stg])
list(res)
```

```python
# count the movies with Meryl Streep
match_stg = {"$match" : {"cast":"Meryl Streep"}}
group_stg = {"$group":{"_id":"null", "count":{"$sum":1}}}
res = db.movies.aggregate([match_stg, group_stg])
list(res)
```

```python
# count the Amerian movies for each rating
match_stg = {"$match" : {"countries":"USA"}}
group_stg = {"$group":{"_id":"$rated", "count":{"$sum":1}}}
res = db.movies.aggregate([match_stg, group_stg])
list(res)
```

```python
# find top 5 Italian movies with the larger cast
# note: in the initial filter I remove the movies for which there is no cast
match_stg = {"$match" : {"countries":"Italy", "cast":{"$exists":"true"}}}
project_stg = {"$project" : {"_id":0, "title":1, "countries":1, "ncast": {"$size":"$cast"}}}
sort_stg = {"$sort": {"ncast":-1}}
limit_stg = {"$limit":5}
res = db.movies.aggregate([match_stg, project_stg, sort_stg, limit_stg])
list(res)
```

```python
target_plot = db.movies.find({"title" : "Casablanca"}, {"_id":0, "title":1, "fullplot":1})[0]['fullplot']
```

```python
df_plot = pd.DataFrame(db.movies.find({ "fullplot": { '$exists': True } }, {"_id":0, "title":1, "fullplot":1}))
```

```python
df_plot
```

```python
plots = list(df_plot['fullplot'])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Step 1: Preprocessing and Feature Extraction using TF-IDF
tfidf_vectorizer = TfidfVectorizer(stop_words='english')  # Remove stop words
X = tfidf_vectorizer.fit_transform(plots)  # Convert text data to TF-IDF vectors
target = tfidf_vectorizer.transform([target_plot])
```

```python
print(X.shape)
print(target.shape)
```

```python
import numpy as np
from scipy.spatial.distance import cosine, euclidean
distances = []

for idx, plot_cand in enumerate(X):
    c = np.squeeze(plot_cand.toarray())
    t = np.squeeze(target.toarray())
    dist = 1 - cosine(c, t)
    dist_e = euclidean(c, t)
    title = df_plot.at[idx, 'title']
    distances.append((dist, title))
```

```python
distances.sort(key=lambda x: x[0], reverse=True)
distances[:6]
```
