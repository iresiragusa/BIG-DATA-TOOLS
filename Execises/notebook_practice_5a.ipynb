Compute distances

```python
# https://docs.scipy.org/doc/scipy/reference/spatial.distance.html

from scipy.spatial.distance import minkowski
from scipy.spatial.distance import chebyshev, cityblock
import math

x=(1,2)
y=(3,4)

print('cityblock', cityblock(x, y))
print('cityblock', minkowski(x, y, 1))
print('euclidean', minkowski(x, y, 2))

# L-1000 vs L-inf
print('L-1000', minkowski(x, y, 1000))
print('L-inf', minkowski(x, y, math.inf))
print('chebyshev', chebyshev(x, y))
```

Compute the match-based similarity, cosine similarity and the Jaccard coefficient, between the two sets {A,B,C} and {A,C,D,E}.

```python
# match-based similarity calculates the proportion of common elements over all elements
# So we have two common elements (A,C) and total five unique elements (A, B, C, D, E).
# Therefore, the match-based similarity is 2/5 = 0.4.
```

```python
# for cosine similarity we can convert sets as one-hot encoding
# set = [a,b,c,d,e]

# s1 = [a,b,c]
s1 = [1, 1, 1, 0, 0]

# s2 = [a,c,d,e]
s2 = [1, 0, 1, 1, 1]

# The cosine similarity is then the dot product of the two vectors
# divided by the product of their norms.

from numpy import dot
from numpy.linalg import norm

print('cosine similarity', dot(s1, s2)/(norm(s1)*norm(s2)))
```

```python
def jaccard_similarity(list1, list2):
    intersection = len(list(set(list1).intersection(list2)))
    union = (len(set(list1)) + len(set(list2))) - intersection
    return float(intersection) / union

s1_l = ['a','b','c']
s2_l = ['a','c','d','e']

print('jaccard', jaccard_similarity(s1_l, s2_l))
```

Compute the Edit distance, and LCSS similarity between (a) ababcabc and babcbc (b) cbacbacba and acbacbacb.

```python
# A Naive recursive Python program to find minimum number
# of operations to convert s1 to s2
def editDistRec(s1, s2, m, n):
    # If first string is empty, the only option is to
    # insert all characters of second string into first
    if m == 0:
        return n

    # If second string is empty, the only option is to
    # remove all characters of first string
    if n == 0:
        return m

    # If last characters of two strings are same, nothing
    # much to do. Get the count for remaining strings.
    if s1[m - 1] == s2[n - 1]:
        return editDistRec(s1, s2, m - 1, n - 1)

    # If last characters are not same, consider all three
    # operations on last character of first string,
    # recursively compute minimum cost for all three
    # operations and take minimum of three values.
    return 1 + min(editDistRec(s1, s2, m, n - 1),    # Insert
                   editDistRec(s1, s2, m - 1, n),    # Remove
                   editDistRec(s1, s2, m - 1, n - 1) # Replace
                   )

def editDist(s1, s2):
    result = editDistRec(s1, s2, len(s1), len(s2))
    print(s1, s2, result)

editDist(s1='ababcabc', s2='babcbc')
editDist(s1='cbacbacba', s2='acbacbacb')
```
